```python
from wordcloud import WordCloud
import matplotlib.pyplot as plt
from collections import Counter
import re
from docx import Document

def read_word_doc(file_path):
    doc = Document(file_path)
    full_text = []
    
    for para in doc.paragraphs:
        full_text.append(para.text)
    
    return "\n".join(full_text)

# Example usage
file_path = r"C:\Users\vineet\Desktop\JD.docx"  # Replace with your file path
document_text = read_word_doc(file_path)



# Clean and preprocess the text
# text_cleaned = re.sub(r'[^a-zA-Z\s]', '', document_text)
text_cleaned = document_text.lower()

# Generate the word cloud
wordcloud = WordCloud(width=800, height=400, background_color='white', colormap='viridis').generate(text_cleaned)

# Display the word cloud
plt.figure(figsize=(10, 5))
plt.imshow(wordcloud, interpolation='bilinear')
plt.axis('off')
plt.show()
```
